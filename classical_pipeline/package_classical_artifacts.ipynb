```python
import os
import shutil

ARTIFACT_DIR = "classical_bilstm_artifacts"
os.makedirs(ARTIFACT_DIR, exist_ok=True)

files_to_save = [
    "classical_model_final.keras",
    "classical_metrics.csv",
    "classical_metrics.npy",
    "classical_histories.npy",
    "tokenizer.pkl"
]

for f in files_to_save:
    if os.path.exists(f):
        shutil.copy(f, ARTIFACT_DIR)

shutil.make_archive(
    ARTIFACT_DIR,
    "zip",
    ARTIFACT_DIR
)

print("[INFO] Data for reproducing this pipeline is stored as classical_bilstm_artifacts.zip")
```

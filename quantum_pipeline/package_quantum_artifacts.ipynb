```python
import shutil
import os

ARTIFACT_DIR = "bilstm_qlstm_artifacts1"
ZIP_NAME = "bilstm_qlstm_artifacts1.zip"

# Remove old zip if exists
if os.path.exists(ZIP_NAME):
    os.remove(ZIP_NAME)

# Create zip in /kaggle/working
shutil.make_archive(
    base_name=ARTIFACT_DIR,
    format="zip",
    root_dir=".",
    base_dir=ARTIFACT_DIR
)

print("\n✅ ZIP CREATED SUCCESSFULLY")
print("📦 Location:", os.path.abspath(ZIP_NAME))
```

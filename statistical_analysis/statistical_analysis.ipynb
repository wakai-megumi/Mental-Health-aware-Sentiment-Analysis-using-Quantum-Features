```python
import numpy as np
import pandas as pd
from scipy import stats

# ============================================================
# Correct Kaggle paths
# ============================================================
CLASSICAL_METRICS_PATH = "/kaggle/input/classicaltokenizer/classical_metrics.csv"
QUANTUM_METRICS_PATH   = "/kaggle/input/quantum-model/qlstm_metrics.csv"

# ============================================================
# Load metrics
# ============================================================
classical_df = pd.read_csv(CLASSICAL_METRICS_PATH)
quantum_df   = pd.read_csv(QUANTUM_METRICS_PATH)

# Extract F1 scores (per run)
f1_classical = classical_df["f1"].values
f1_quantum   = quantum_df["f1"].values

print("Classical F1 per run:", f1_classical)
print("Quantum F1 per run  :", f1_quantum)

assert len(f1_classical) == len(f1_quantum), "Run count mismatch!"

# ============================================================
# Paired t-test
# ============================================================
t_stat, p_value = stats.ttest_rel(f1_quantum, f1_classical)

print("\nPaired t-test results:")
print(f"t-statistic = {t_stat:.4f}")
print(f"p-value     = {p_value:.6f}")

# ============================================================
# 95% Confidence Interval for mean difference
# ============================================================
diff = f1_quantum - f1_classical
mean_diff = diff.mean()
std_diff  = diff.std(ddof=1)
n = len(diff)

ci_low, ci_high = stats.t.interval(
    0.95,
    df=n - 1,
    loc=mean_diff,
    scale=std_diff / np.sqrt(n)
)

print("\nF1-score difference (Quantum − Classical):")
print(f"Mean difference = {mean_diff:.6f}")
print(f"95% CI          = [{ci_low:.6f}, {ci_high:.6f}]")

# ============================================================
# Effect size (Cohen's d)
# ============================================================
cohens_d = mean_diff / std_diff
print(f"Cohen's d       = {cohens_d:.4f}")
```
